```python
#yeild example to print even numbers from a list

def print_even(test_list) :
    for i in test_list:
        if i % 2 == 0:
            yield i

test_list = [1, 4, 5, 6, 7]

print ("The original list is : " + str(test_list))

# printing even numbers
print ("The even numbers in list are : ", end = " ")
for j in print_even(test_list):
    print (j, end = " ")
```

```python
# Singly Linked List implementation in Python

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    def append(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            return
        curr = self.head
        while curr.next:
            curr = curr.next
        curr.next = new_node

    def prepend(self, data):
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    def delete(self, key):
        curr = self.head

        if curr and curr.data == key:
            self.head = curr.next
            return True

        prev = None
        while curr and curr.data != key:
            prev = curr
            curr = curr.next

        if not curr:
            return False

        prev.next = curr.next
        return True

    def search(self, key):
        curr = self.head
        while curr:
            if curr.data == key:
                return True
            curr = curr.next
        return False

    def to_list(self):
        result = []
        curr = self.head
        while curr:
            result.append(curr.data)
            curr = curr.next
        return result


# Example usage
ll = LinkedList()
ll.append(10)
ll.append(20)
ll.prepend(5)
ll.append(30)

print("Linked list:", ll.to_list())      # [5, 10, 20, 30]
print("Search 20:", ll.search(20))       # True
print("Delete 10:", ll.delete(10))       # True
print("After delete:", ll.to_list())     # [5, 20, 30]
```

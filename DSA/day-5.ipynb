```python
import heapq  # Provides an efficient heap-based priority queue implementation


class PriorityQueue:  # Defines a priority queue wrapper around a min-heap
    def __init__(self):  # Runs when a new PriorityQueue object is created
        self._heap = []  # Stores items as (priority, value) pairs

    def push(self, item, priority):  # Adds a new item with its priority
        heapq.heappush(self._heap, (priority, item))  # Inserts into the heap

    def pop(self):  # Removes and returns the item with the smallest priority
        if not self._heap:  # Checks whether the queue is empty
            raise IndexError("pop from empty priority queue")  # Raises an error if empty
        return heapq.heappop(self._heap)[1]  # Removes smallest priority and returns the item

    def peek(self):  # Returns the item with the smallest priority without removing it
        if not self._heap:  # Checks whether the queue is empty
            raise IndexError("peek from empty priority queue")  # Raises an error if empty
        return self._heap[0][1]  # Returns the item at the top of the heap

    def is_empty(self):  # Checks whether the queue has no items
        return len(self._heap) == 0  # Returns True if the heap is empty

    def __len__(self):  # Allows len(priority_queue) to work
        return len(self._heap)  # Returns the number of stored items


# Example usage
pq = PriorityQueue()  # Creates a new priority queue
pq.push("low", 5)  # Adds an item with priority 5
pq.push("high", 1)  # Adds an item with priority 1
pq.push("medium", 3)  # Adds an item with priority 3

print(pq.pop())   # Outputs: high
print(pq.peek())  # Outputs: medium
print(len(pq))    # Outputs: 2
```

what is hash


```python
A hash data structure, often called a **hash table** or **hash map**, is a way to store and find data very quickly.

## Basic idea
It stores data as **key-value pairs**:
- **key** = the name you use to look up data
- **value** = the data itself

Example:
- `"name"` → `"Alice"`
- `"age"` → `25`

## Why it is useful
Instead of searching through everything one by one, a hash table uses a **hash function** to turn a key into an index, which makes lookup very fast.

## Simple analogy
Think of a hash table like a library system:
- the **book title** is the key
- the **shelf location** is found quickly using a rule
- you do not check every shelf manually

## Main operations
- **Insert**: add a key-value pair
- **Search**: find a value using a key
- **Delete**: remove a key-value pair

## In Python
The built-in `dict` is a hash table:
- `person["name"]`
- `person["age"]`

## Important idea
Hash tables are usually very fast, but sometimes two keys can produce the same location. This is called a **collision**. Good hash tables have ways to handle collisions.

## Short definition
A hash data structure is a fast way to store and retrieve data using keys.
```

```python

class SimpleHashMap:
    def __init__(self, size=10):
        self.size = size
        self.table = [[] for _ in range(size)]  # list of buckets

    def _index(self, key):
        return hash(key) % self.size  # map key to bucket index

    def set(self, key, value):
        idx = self._index(key)
        bucket = self.table[idx]

        # update value if key already exists
        for i, (k, v) in enumerate(bucket):
            if k == key:
                bucket[i] = (key, value)
                return

        # otherwise add new pair
        bucket.append((key, value))

    def get(self, key):
        idx = self._index(key)
        bucket = self.table[idx]

        for k, v in bucket:
            if k == key:
                return v
        return None  # key not found

    def delete(self, key):
        idx = self._index(key)
        bucket = self.table[idx]

        for i, (k, v) in enumerate(bucket):
            if k == key:
                del bucket[i]
                return True
        return False


### Example usage


h = SimpleHashMap()

h.set("name", "Alice")
h.set("age", 25)

print(h.get("name"))  # Alice
print(h.get("age"))   # 25

h.delete("age")
print(h.get("age"))   # None
```

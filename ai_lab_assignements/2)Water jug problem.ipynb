# Depth First Search (DFS)

```python
# DFS for Water Jug Problem (With Process Display)
def dfs(capacity1, capacity2, target):
    # Stack to store the current state (x, y)
    stack = [(0, 0)]  # Start with both jugs empty
    visited = set()  # To avoid revisiting states
    actions = {  # Action descriptions for each move
        "fill_1": "Fill Jug 1",
        "fill_2": "Fill Jug 2",
        "empty_1": "Empty Jug 1",
        "empty_2": "Empty Jug 2",
        "pour_1_to_2": "Pour Jug 1 into Jug 2",
        "pour_2_to_1": "Pour Jug 2 into Jug 1"
    }

    while stack:
        state = stack.pop()  # Pop the last state
        x, y = state

        if (x, y) == target:
            print(f"Solution found: ({x}, {y})")
            return True  # Target reached

        if (x, y) in visited:
            continue  # Skip if the state has already been visited
        
        visited.add((x, y))  # Mark the current state as visited

        # Possible actions to take from the current state
        next_states = [
            (capacity1, y, "fill_1"),  # Fill Jug 1
            (x, capacity2, "fill_2"),  # Fill Jug 2
            (0, y, "empty_1"),  # Empty Jug 1
            (x, 0, "empty_2"),  # Empty Jug 2
            (x - min(x, capacity2 - y), y + min(x, capacity2 - y), "pour_1_to_2"),  # Pour Jug 1 to Jug 2
            (x + min(y, capacity1 - x), y - min(y, capacity1 - x), "pour_2_to_1")   # Pour Jug 2 to Jug 1
        ]

        # Add the valid next states to the stack
        for next_state in next_states:
            new_x, new_y, action = next_state
            if (new_x, new_y) not in visited:
                stack.append((new_x, new_y))
                print(f"Action: {actions[action]} -> State: ({new_x}, {new_y})")

    print("No solution found.")
    return False  # No solution found

# Example usage
capacity1 = 4  # Capacity of Jug 1
capacity2 = 3  # Capacity of Jug 2
target = (2, 0)  # Target amount in Jug 1 is 2 liters
print("Solving using DFS:")
dfs(capacity1, capacity2, target)
```

# Breadth First Search

```python
from collections import deque

# BFS for Water Jug Problem (With Process Display)
def bfs(capacity1, capacity2, target):
    # Queue for BFS (First-In, First-Out)
    queue = deque([(0, 0)])  # Start with both jugs empty
    visited = set()  # To avoid revisiting states
    actions = {  # Action descriptions for each move
        "fill_1": "Fill Jug 1",
        "fill_2": "Fill Jug 2",
        "empty_1": "Empty Jug 1",
        "empty_2": "Empty Jug 2",
        "pour_1_to_2": "Pour Jug 1 into Jug 2",
        "pour_2_to_1": "Pour Jug 2 into Jug 1"
    }

    while queue:
        state = queue.popleft()  # Pop the first state in the queue
        x, y = state

        if (x, y) == target:
            print(f"Solution found: ({x}, {y})")
            return True  # Target reached

        if (x, y) in visited:
            continue  # Skip if the state has already been visited
        
        visited.add((x, y))  # Mark the current state as visited

        # Possible actions to take from the current state
        next_states = [
            (capacity1, y, "fill_1"),  # Fill Jug 1
            (x, capacity2, "fill_2"),  # Fill Jug 2
            (0, y, "empty_1"),  # Empty Jug 1
            (x, 0, "empty_2"),  # Empty Jug 2
            (x - min(x, capacity2 - y), y + min(x, capacity2 - y), "pour_1_to_2"),  # Pour Jug 1 to Jug 2
            (x + min(y, capacity1 - x), y - min(y, capacity1 - x), "pour_2_to_1")   # Pour Jug 2 to Jug 1
        ]

        # Add the valid next states to the queue
        for next_state in next_states:
            new_x, new_y, action = next_state
            if (new_x, new_y) not in visited:
                queue.append((new_x, new_y))
                print(f"Action: {actions[action]} -> State: ({new_x}, {new_y})")

    print("No solution found.")
    return False  # No solution found

# Example usage
capacity1 = 4  # Capacity of Jug 1
capacity2 = 3  # Capacity of Jug 2
target = (2, 0)  # Target amount in Jug 1 is 2 liters
print("Solving using BFS:")
bfs(capacity1, capacity2, target)
```

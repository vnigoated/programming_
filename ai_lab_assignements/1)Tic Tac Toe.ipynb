```python
import random
import time
```

```python
# Constants for the game
COMPUTER = 1
HUMAN = 2
SIDE = 3
COMPUTERMOVE = 'O'
HUMANMOVE = 'X'
```

```python
# Function to initialise the game / Tic-Tac-Toe board
def initialise():
    board = [[' ' for _ in range(SIDE)] for _ in range(SIDE)]
    moves_available = [(i+1) for i in range(SIDE*SIDE)]
    moves_played = []
    return board, moves_available, moves_played
```

```python
# Function to print the Tic-Tac-Toe board
def showBoard(board):
    print("\n\n")
    print("\t\t\t  {} | {}  | {}  ".format(board[0][0], board[0][1], board[0][2]))
    print("\t\t\t--------------")
    print("\t\t\t  {} | {}  | {}  ".format(board[1][0], board[1][1], board[1][2]))
    print("\t\t\t--------------")
    print("\t\t\t  {} | {}  | {}  \n\n".format(board[2][0], board[2][1], board[2][2]))
```

```python
# Function to show the instructions
def showInstructions():
    print("\t\t\t  Tic-Tac-Toe\n\n")
    print("Choose a cell numbered from 1 to 9 as below and play\n\n")
    print("\t\t\t  1 | 2  | 3  ")
    print("\t\t\t--------------")
    print("\t\t\t  4 | 5  | 6  ")
    print("\t\t\t--------------")
    print("\t\t\t  7 | 8  | 9  \n\n")
    print("-\t-\t-\t-\t-\t-\t-\t-\t-\t-\n\n")
```

```python
# Function to declare the winner of the game
def declareWinner(whoseTurn):
    if whoseTurn == COMPUTER:
        print("COMPUTER has won")
    else:
        print("HUMAN has won")
```

```python
# Functions to check if any of the rows, columns, or diagonals have been crossed
def rowCrossed(board):
    for i in range(SIDE):
        if board[i][0] == board[i][1] and board[i][1] == board[i][2] and board[i][0] != ' ':
            return True
    return False

def columnCrossed(board):
    for i in range(SIDE):
        if board[0][i] == board[1][i] and board[1][i] == board[2][i] and board[0][i] != ' ':
            return True
    return False

def diagonalCrossed(board):
    if board[0][0] == board[1][1] and board[1][1] == board[2][2] and board[0][0] != ' ':
        return True
    if board[0][2] == board[1][1] and board[1][1] == board[2][0] and board[0][2] != ' ':
        return True
    return False
```

```python
# Function to check if the game is over
def gameOver(board):
    return rowCrossed(board) or columnCrossed(board) or diagonalCrossed(board)
```

```python
# Function to play Tic-Tac-Toe
def playTicTacToe(whoseTurn):
    board, moves_available, moves_played = initialise()
    showInstructions()
    while not gameOver(board) and moves_available != []:
        if whoseTurn == COMPUTER:                       #Computer moves
            while whoseTurn == COMPUTER:
                move = random.randint(1,9)
                if (move in moves_available):
                    moves_available.remove(move)
                    moves_played.append(move)
                    move -= 1
                    x = move // SIDE
                    y = move % SIDE
                    board[x][y] = COMPUTERMOVE
                    print("COMPUTER has put a {} in cell {}".format(COMPUTERMOVE, move + 1))
                    showBoard(board)
                    whoseTurn = HUMAN
                else:
                    move = random.randint(1,9)
            print ("Moves available: ",moves_available)
            print ("Moves played: ",moves_played)
        elif whoseTurn == HUMAN:                        #Human moves
            while whoseTurn == HUMAN:
                move = int(input("Select position:"))
                if (move in moves_available):
                    moves_available.remove(move)
                    moves_played.append(move)
                    move -= 1
                    x = move // SIDE
                    y = move % SIDE
                    board[x][y] = HUMANMOVE
                    print("HUMAN has put a {} in cell {}".format(HUMANMOVE, move + 1))
                    showBoard(board)
                    whoseTurn = COMPUTER
                else:
                    move = int(input("Move not available. Select another position:"))
                print ("Moves available: ",moves_available)
                print ("Moves played: ",moves_played)
    if not gameOver(board) and moves_available == []:
        print("It's a draw")
    else:
        if whoseTurn == COMPUTER:
            whoseTurn = HUMAN
        elif whoseTurn == HUMAN:
            whoseTurn = COMPUTER
        declareWinner(whoseTurn)
```

```python
# Driver function
if __name__ == "__main__":
    playTicTacToe(HUMAN) #Change HUMAN to COMPUTER for Computer to play first move
```
